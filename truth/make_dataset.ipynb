```python
import pandas as pd

pos_df = pd.read_csv('./data/cities.csv')
pos_true = pos_df[pos_df["label"]==1]
pos_false = pos_df[pos_df["label"]==0]

neg_df = pd.read_csv('./data/neg_cities.csv')
neg_true = neg_df[neg_df["label"]==1]
neg_false = neg_df[neg_df["label"]==0]

true = pd.concat([pos_true, neg_true], ignore_index=True)
false = pd.concat([pos_false, neg_false], ignore_index=True)
```

```python
import json

statements = [
    {
        "true":  true["statement"].iloc[i],
        "false": false["statement"].iloc[i]
    }
    for i in range(len(true))
]


with open("./data/true_false.jsonl", "w") as f:
    for statement in statements:
        f.write(json.dumps(statement) + "\n")
```
